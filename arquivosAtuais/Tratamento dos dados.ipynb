```python
import pandas as pd

arquivo = "planilha.xlsx"

df = pd.read_excel(arquivo, sheet_name="Sheet1")

excecoes = {
    "de", "da", "do", "das", "dos",
    "e", "em", "no", "na", "nos", "nas",
    "para", "por", "com", "a", "o", "as", "os",
    "à", "às", "ao", "aos"
}

def capitalizar_titulo(texto):
    if pd.isna(texto):
        return texto

    palavras = str(texto).strip().split()

    resultado = []
    for i, palavra in enumerate(palavras):
        palavra = palavra.lower()

        if i == 0 or palavra not in excecoes:
            resultado.append(palavra.capitalize())
        else:
            resultado.append(palavra)

    return " ".join(resultado)

# Coluna B (índice 1)
df.iloc[:, 1] = df.iloc[:, 1].apply(capitalizar_titulo)

# Sobrescreve o arquivo original
df.to_excel(arquivo, index=False)

print("Planilha atualizada com sucesso!")
```

```python
from openpyxl import load_workbook

arquivo = "planilha.xlsx"

wb = load_workbook(arquivo)
ws = wb["Sheet1"]

excecoes = {
    "de", "da", "do", "das", "dos",
    "e", "em", "no", "na", "nos", "nas",
    "para", "por", "com", "a", "o", "as", "os",
    "à", "às", "ao", "aos"
}

def capitalizar_titulo(texto):
    if texto is None:
        return texto

    palavras = str(texto).strip().split()

    resultado = []
    for i, palavra in enumerate(palavras):
        palavra = palavra.lower()

        if i == 0 or palavra not in excecoes:
            resultado.append(palavra.capitalize())
        else:
            resultado.append(palavra)

    return " ".join(resultado)

# Coluna B
for linha in range(2, ws.max_row + 1):
    celula = ws[f'B{linha}']
    if celula.value:
        celula.value = capitalizar_titulo(celula.value)

# Coluna C
for linha in range(2, ws.max_row + 1):
    celula = ws[f'C{linha}']
    if celula.value:
        celula.value = capitalizar_titulo(celula.value)

# Colunas D, E e F
for coluna in ['D', 'E', 'F']:
    for linha in range(2, ws.max_row + 1):
        celula = ws[f'{coluna}{linha}']
        if celula.value:
            celula.value = str(celula.value).title()

wb.save(arquivo)

print("Planilha atualizada mantendo toda a formatação.")
```

```python
import re
from openpyxl import load_workbook

arquivo = "planilha.xlsx"

wb = load_workbook(arquivo)
ws = wb["Sheet1"]

for linha in range(2, ws.max_row + 1):
    celula = ws[f'H{linha}']

    if celula.value is None or str(celula.value).strip() == "":
        celula.value = 1
    else:
        texto = str(celula.value)

        numeros = re.findall(r'\d+', texto)

        if numeros:
            celula.value = int(numeros[0])
        else:
            celula.value = 1

wb.save(arquivo)

print("Coluna H atualizada.")
```

```python
import re
from openpyxl import load_workbook

arquivo = "planilha.xlsx"

wb = load_workbook(arquivo)
ws = wb["Sheet1"]

def normalizar_cdd(valor):
    if valor is None:
        return ""

    texto = str(valor).strip().upper()

    # Guarda a primeira letra, se houver
    letra = ""
    m = re.match(r'^([A-Z])', texto)
    if m:
        letra = m.group(1)

    # Remove CDU e textos entre parênteses
    texto = re.sub(r'\(.*?\)', '', texto)

    # Extrai apenas os dígitos
    digitos = ''.join(re.findall(r'\d', texto))

    if not digitos:
        return letra

    if len(digitos) <= 3:
        return letra + digitos

    inteiro = digitos[:3]
    decimal = digitos[3:7]  # até 4 casas

    return letra + inteiro + "." + decimal

for linha in range(2, ws.max_row + 1):
    celula = ws[f'K{linha}']

    if celula.value:
        celula.value = normalizar_cdd(celula.value)

wb.save(arquivo)

print("Coluna K atualizada.")
```

```python
!pip install pymarc openpyxl -q

import datetime
from pymarc import Record, Field, MARCWriter, Subfield
from openpyxl import load_workbook

arquivo_excel = "planilha.xlsx"

wb = load_workbook(arquivo_excel)
ws = wb["Sheet1"]

with open("biblioteca.mrc", "wb") as output:
    writer = MARCWriter(output)

    # Processando as linhas do Excel (linhas 3 e 4)
    for linha in range(3, 5):

        titulo = ws[f'B{linha}'].value
        subtitulo = ws[f'C{linha}'].value
        autor = ws[f'D{linha}'].value
        editora = ws[f'E{linha}'].value
        cidade = ws[f'F{linha}'].value
        ano = ws[f'G{linha}'].value
        edicao = ws[f'H{linha}'].value
        assuntos = ws[f'I{linha}'].value
        palavras = ws[f'J{linha}'].value
        cdd = ws[f'K{linha}'].value

        # Se não houver título, ignora a linha
        if not titulo:
            continue

        record = Record(force_utf8=True)

        # --- CAMPOS DE CONTROLE (Usa add_field com o argumento 'data') ---

        # 001 - ID do registro (único)
        record.add_field(Field(tag='001', data=f'{linha:07d}'))

        # 005 - Data e hora da transação (Formato exato do Biblivre)
        agora = datetime.datetime.now().strftime("%Y%m%d%H%M%S.0")
        record.add_field(Field(tag='005', data=agora))

        # 008 - Dados fixos (40 caracteres exatos baseados na sua exportação)
        data_008 = datetime.datetime.now().strftime("%y%m%d")
        ano_pub = str(ano)[:4] if ano else "||||"
        # Monta rigorosamente os 40 caracteres exigidos pelo Biblivre
        campo_008_str = f"{data_008}s{ano_pub}    bl|||||||||||||||||por|u"
        record.add_field(Field(tag='008', data=campo_008_str))

        # --- CAMPOS DE DADOS ---

        # CDD (082)
        if cdd:
            record.add_field(
                Field(
                    tag='082',
                    indicators=['0', '4'],
                    subfields=[Subfield('a', str(cdd))]
                )
            )

        # Autor (100)
        if autor:
            record.add_field(
                Field(
                    tag='100',
                    indicators=['1', ' '], # Espaço em branco real para indicador vazio
                    subfields=[Subfield('a', str(autor))]
                )
            )

        # Título (245)
        subfields_245 = [Subfield('a', str(titulo))]
        if subtitulo:
            subfields_245.append(Subfield('b', f" : {str(subtitulo)}"))
        if autor:
            subfields_245.append(Subfield('c', f" / {str(autor)}"))

        record.add_field(
            Field(
                tag='245',
                indicators=['1', '0'],
                subfields=subfields_245
            )
        )

        # Edição (250)
        if edicao:
            record.add_field(
                Field(
                    tag='250',
                    indicators=[' ', ' '],
                    subfields=[Subfield('a', str(edicao))]
                )
            )

        # Publicação (260)
        subfields_260 = []
        if cidade:
            subfields_260.append(Subfield('a', str(cidade)))
        if editora:
            prefixo_ed = " : " if cidade else ""
            subfields_260.append(Subfield('b', f"{prefixo_ed}{str(editora)}"))
        if ano:
            prefixo_ano = ", " if (editora or cidade) else ""
            subfields_260.append(Subfield('c', f"{prefixo_ano}{str(ano)}"))

        if subfields_260:
            record.add_field(
                Field(
                    tag='260',
                    indicators=[' ', ' '],
                    subfields=subfields_260
                )
            )

        # Assuntos (650)
        assuntos_adicionados = set()

        def adicionar_650(lista_termos):
            if lista_termos:
                for termo in str(lista_termos).split(';'):
                    termo = termo.strip()
                    if termo and termo not in assuntos_adicionados:
                        assuntos_adicionados.add(termo)
                        record.add_field(
                            Field(
                                tag='650',
                                indicators=[' ', '4'],
                                subfields=[Subfield('a', termo)]
                            )
                        )

        adicionar_650(assuntos)
        adicionar_650(palavras)

        # Salva o registro estruturado no arquivo
        writer.write(record)

    writer.close()

print("Arquivo biblioteca.mrc gerado com sucesso!")
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from openpyxl import load_workbook
import time

# PLANILHA
wb = load_workbook("planilha.xlsx")
ws = wb["Sheet1"]

# CHROME
options = webdriver.ChromeOptions()

driver = webdriver.Chrome(options=options)

print("[INFO] Abrindo Biblivre...")
driver.get("http://localhost/Biblivre5/?action=cataloging_bibliographic")

time.sleep(3)

print("[INFO] Página carregada")

for linha in range(3, 5):

    print("\n" + "=" * 60)
    print(f"[LINHA {linha}] Iniciando processamento")

    titulo = ws[f'B{linha}'].value or ""
    subtitulo = ws[f'C{linha}'].value or ""
    autor = ws[f'D{linha}'].value or ""
    editora = ws[f'E{linha}'].value or ""
    cidade = ws[f'F{linha}'].value or ""
    ano = ws[f'G{linha}'].value or ""
    assuntos = ws[f'I{linha}'].value or ""
    cdd = ws[f'K{linha}'].value or ""

    print(f"[LINHA {linha}] Título: {titulo}")

    lista_assuntos = [
        x.strip()
        for x in str(assuntos).split(";")
        if x.strip()
    ]

    primeiro_assunto = lista_assuntos[0] if lista_assuntos else ""
    ultimo_assunto = lista_assuntos[-1] if lista_assuntos else ""

    try:

        print(f"[LINHA {linha}] Clicando em Novo Registro")

        driver.find_element(
            By.ID,
            "new_record_button"
        ).click()

        time.sleep(2)

        driver.save_screenshot(
            f"linha_{linha}_01_novo_registro.png"
        )

        ativo = driver.switch_to.active_element

        print(f"[LINHA {linha}] Preenchendo CDD")

        for _ in range(4):
            ativo.send_keys(Keys.TAB)

        ativo.send_keys(str(cdd))

        print(f"[LINHA {linha}] Preenchendo Autor")

        for _ in range(8):
            ativo.send_keys(Keys.TAB)

        ativo.send_keys(str(autor))

        print(f"[LINHA {linha}] Preenchendo Título")

        for _ in range(20):
            ativo.send_keys(Keys.TAB)

        ativo.send_keys(str(titulo))

        print(f"[LINHA {linha}] Preenchendo Subtítulo")

        ativo.send_keys(Keys.TAB)
        ativo.send_keys(str(subtitulo))

        print(f"[LINHA {linha}] Preenchendo Cidade")

        for _ in range(13):
            ativo.send_keys(Keys.TAB)

        ativo.send_keys(str(cidade))

        print(f"[LINHA {linha}] Preenchendo Editora")

        for _ in range(2):
            ativo.send_keys(Keys.TAB)

        ativo.send_keys(str(editora))

        print(f"[LINHA {linha}] Preenchendo Ano")

        for _ in range(2):
            ativo.send_keys(Keys.TAB)

        ativo.send_keys(str(ano))

        print(f"[LINHA {linha}] Preenchendo Primeiro Assunto")

        for _ in range(7):
            ativo.send_keys(Keys.TAB)

        ativo.send_keys(primeiro_assunto)

        print(f"[LINHA {linha}] Preenchendo Último Assunto")

        for _ in range(9):
            ativo.send_keys(Keys.TAB)

        ativo.send_keys(ultimo_assunto)

        driver.save_screenshot(
            f"linha_{linha}_02_preenchida.png"
        )

        time.sleep(1)

        print(f"[LINHA {linha}] Clicando em Salvar")

        driver.find_element(
            By.XPATH,
            "//a[contains(@onclick,'saveRecord')]"
        ).click()

        time.sleep(3)

        driver.save_screenshot(
            f"linha_{linha}_03_salva.png"
        )

        print(f"[LINHA {linha}] Registro salvo")

    except Exception as erro:

        print(f"[ERRO LINHA {linha}] {erro}")

        driver.save_screenshot(
            f"linha_{linha}_erro.png"
        )

        break

print("\n[INFO] Processo finalizado")
```
